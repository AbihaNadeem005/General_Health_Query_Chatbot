## **Step 3: Chatbot Function**
This function sends user input to the model using prompt engineering and returns a generated response.

```python
def health_chatbot(user_input):
    prompt = f"""
Act as a helpful and friendly medical assistant.

Rules:
- Give general health information only
- Do not give prescriptions or diagnosis
- Suggest consulting a doctor if needed

Question: {user_input}

Answer:
"""

    response = chatbot(
        prompt,
        max_new_tokens=150,
        do_sample=True
    )

    result = response[0]['generated_text']

    # clean output
    answer = result.split("Answer:")[-1].strip()
    # remove repeated prompt artifacts
    if "Question:" in answer:
     answer = answer.split("Question:")[0].strip()

    return answer
```

## **Step 4: Safety Filter**
This function blocks harmful or unsafe medical queries like self-harm or overdose.

```python
def is_safe_query(text):
    danger_words = [
        "suicide", "self harm", "overdose", "drug dose",
        "poison", "kill myself", "lethal dose", "fatal dose", "how to die", "death dose"
    ]

    text = text.lower()
    return not any(word in text for word in danger_words)
```

## **Step 5: Interactive Loop**
This section runs the chatbot in an interactive mode. It continuously takes user input, checks it through a safety filter, and then generates a response using the LLM. The loop stops when the user types 'exit'.

```python
print("Health Chatbot is running... (type 'exit' to stop)\n")

while True:
    user = input("Ask a health question: ")

    if user.lower() == "exit":
        print("Goodbye! Stay healthy")
        break

    if not is_safe_query(user):
        print("Please consult a healthcare professional.\n")
    else:
        print(health_chatbot(user))
        print("\n")
```
